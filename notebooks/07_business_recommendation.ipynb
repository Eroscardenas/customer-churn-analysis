## From Churn Prediction to Retention Strategy

This notebook translates the analytical and predictive results into actionable business recommendations.

Rather than focusing only on model performance, the objective is to identify the main drivers of customer churn and define practical strategies that could improve customer retention.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load model evaluation results
results = pd.read_csv("../reports/model_evaluation.csv")

results.round(3)
```

```python
# Load Gradient Boosting feature importance
feature_importance = pd.read_csv(
    "../reports/gradient_boosting_feature_importance.csv"
)

top_features = feature_importance.head(8)

top_features
```

```python
plt.figure(figsize=(8, 5))

plt.barh(
    top_features["Feature"][::-1],
    top_features["Importance"][::-1]
)

plt.title("Top Churn Drivers — Gradient Boosting")
plt.xlabel("Feature Importance")
plt.ylabel("")

plt.tight_layout()
plt.show()
```

## Key Churn Drivers

The predictive analysis highlights a small set of factors strongly associated with customer churn:

- **Contract type:** Month-to-month customers represent a key churn-risk segment.
- **Customer tenure:** Newer customers are more vulnerable to leaving.
- **Charges:** Monthly and total charges contribute significantly to churn prediction.
- **Fiber optic service:** Customers using fiber optic internet show a relevant churn pattern.
- **Support services:** Customers without Online Security or Tech Support appear more likely to churn.

These findings suggest that retention efforts should focus on customer lifecycle, contract commitment, pricing perception, and support-service adoption.

## High-Risk Customer Profile

Based on the exploratory analysis and predictive models, customers with higher churn risk tend to share the following characteristics:

- Month-to-month contracts
- Short customer tenure
- Higher monthly charges
- Fiber optic internet service
- No Online Security
- No Tech Support
- Electronic check as payment method

This profile provides a practical way to identify customer segments that should receive retention attention first.

## Business Recommendations

| Priority | Recommendation | Target |
|---|---|---|
| High | Encourage migration from month-to-month to longer-term contracts through loyalty incentives. | Month-to-month customers |
| High | Create early-retention campaigns during the first months of the customer lifecycle. | New customers |
| High | Promote Tech Support and Online Security as retention-oriented service bundles. | Customers without support services |
| Medium | Review pricing and perceived value among customers with high monthly charges. | High-charge customers |
| Medium | Implement proactive retention campaigns using churn-risk predictions. | High-risk customers |
